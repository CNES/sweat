# Extract EC data

## Input

```python
input_data_path = os.path.join(
    os.path.expanduser("~"), "data", "validation", "Data.csv"
)
ameriflux_base_dir = os.path.join(
    os.path.expanduser("~"), "data", "ameriflux", "base"
)
fluxnet_base_dir = os.path.join(
    os.path.expanduser("~"), "data", "ameriflux", "fluxnet"
)
```

## Read raw data

```python
input_data = pr.read_input_data(input_data_path)
```

```python
input_data.head()
```

```python
input_data.columns
```

## Prepare data

* Compute local time
* Compute datetime (in local time)



```python
data = pr.prepare_input_data(input_data)
```

```python
data.head()
```

## Get Amiflux base data

```python
_, am_data = pr.get_ameriflux_data(data, ameriflux_base_dir)
```

## Get Fluxnet data

```python
_, fl_data = pr.get_fluxnet_data(data, fluxnet_base_dir)
```

## Merge

```python
complete_data = pd.merge(left=data, right=am_data, on=["date", "name"])
complete_data = pd.merge(left=complete_data, right=fl_data, on=["date", "name"])
```

```python
complete_data.head()
```

## Write CSV results

```python
complete_data.to_csv(
    os.path.join(
        os.path.expanduser("~"), "data", "validation", "data_ameriflux.csv"
    ),
    index=False,
)
```
