Declination angle

```python
days = np.arange(1, 366)
dates = np.array(
    [dt.date(2025, 1, 1) + dt.timedelta(days=d) for d in range(365)]
)
```

```python
decl_angle = np.rad2deg([solar._declination_angle(date) for date in dates])
```

```python
decl_angle_2 = 23.45 * np.sin(2 * np.pi * (days + 284) / 365.0)
```

```python
plt.plot(days, decl_angle, color="blue", label="Spencer71")
plt.plot(days, decl_angle_2, color="red", label="Cooper69")
plt.xlabel("day of year")
plt.ylabel("declination angle (degree)")
plt.legend()
```

## Solar irradiance

TOA solar radiation

```python
dates = [dt.date(2025, 1, 1) + dt.timedelta(i) for i in range(365)]
solar_radiation = [
    solar.compute_toa_solar_radiation_from_sun_angles(d, sza=0, saa=0)
    for d in dates
]
approx_solar_radiation = [
    solar.SOLAR_FLUX * (1 + 0.033 * np.cos(2 * np.pi * (d - 2) / 365.0))
    for d in range(365)
]
```

```python
plt.plot(
    dates, solar_radiation, color="blue", label="Solar radiation (Spencer)"
)
plt.plot(
    dates, approx_solar_radiation, color="red", label="Solar radiation (approx)"
)
plt.xlabel("day of year")
plt.ylabel("Solar radiation (W.m-2)")
plt.legend()
```

```python
date = dt.datetime(2025, 1, 9, 11, 30, 0, tzinfo=dt.UTC)
sza, saa = solar.compute_sun_angles(date, y=43.0, x=1.0)
print(f"SZA = {sza}")
print(f"SAA = {saa}")
toa = solar.compute_toa_solar_radiation_from_sun_angles(date, sza, saa)
print(toa)
```

TOA daily solar radiation

```python
dates = [dt.date(2025, 1, 1) + dt.timedelta(i) for i in range(365)]
solar_radiation = [
    solar.compute_daily_toa_solar_radiation(d, y=43.0, x=1.0) for d in dates
]
solar_radiation_ha = [
    solar.compute_daily_toa_solar_radiation_from_hour_angle(d, y=43.0, x=1.0)
    for d in dates
]
solar_radiation_eq = [
    solar._toa_daily_irradiance(d, y=43.0, x=1.0) for d in dates
]
```

```python
plt.plot(dates, solar_radiation, label="num")
plt.plot(dates, solar_radiation_ha, label="num_ha")
plt.plot(dates, solar_radiation_eq, label="eq")
plt.xlabel("day of year")
plt.ylabel("Daily solar radiation (kJ.m-2.day-1)")
plt.legend()
```
