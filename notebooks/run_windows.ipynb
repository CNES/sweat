```python
et_single_date_dir = "/home/prohel/Bureau/trishna/et-dataset/timeseries/et_copy"
radiation_dir = (
    "/home/prohel/Bureau/trishna/et-dataset/timeseries/daily_radiation"
)
et_time_series_dir = "/home/prohel/Bureau/trishna/sweat/out/et_time_series_dir/"
json_dir = "/home/prohel/Bureau/trishna/sweat/out/json_dir"
```

### Generating start and end dates for each sliding window shift within the given period (a 7-day window with a 1-day shift)

```python
a = window_generator(min_date_dt, max_date_dt, 7, 1)
```
