## Solar position

Declination angle

```python
days = np.arange(1, 365)
```

```python
decl_angle = daily._declination_angle(days)
```

```python
decl_angle_2 = 23.45 * np.sin(2 * np.pi * (days + 284) / 365.0)
```

```python
plt.plot(days, decl_angle, color="blue", label="Spencer71")
plt.plot(days, decl_angle_2, color="red", label="Cooper69")
plt.xlabel("day of year")
plt.ylabel("declination angle (degree)")
plt.legend()
```

## Solar irradiance

TOA solar radiation

```python
days = np.arange(1, 365)
solar_radiation = daily.toa_instant_irradiance(days, sza=0)
approx_solar_radiation = daily.SOLAR_FLUX * (
    1 + 0.033 * np.cos(2 * np.pi * (days - 2) / 365.0)
)
```

```python
plt.plot(days, solar_radiation, color="blue", label="Solar radiation (Spencer)")
plt.plot(
    days, approx_solar_radiation, color="red", label="Solar radiation (approx)"
)
plt.xlabel("day of year")
plt.ylabel("Solar radiation (W.m-2)")
plt.legend()
```

```python
d = daily._declination_angle(9)
print(d)
h = daily._hour_angle(
    dt.datetime(2025, 1, 9, 11, 30, 0, tzinfo=dt.timezone.utc),
    lon=1.0,
    lat=43.0,
)
print(h)
sza = daily._sun_zenith_angle(lat=43.0, hour_angle=h, decl_angle=d)
print(sza)
toa = daily.toa_instant_irradiance(9, sza)
print(toa)
```

TOA daily solar radiation

```python
days = np.arange(1, 365)
solar_radiation = daily.toa_daily_irradiance(days, lat=43.0)
```
