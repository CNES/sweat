### Extract and set a bit

```python
status = xr.DataArray(sh.STATUS_TYPE(849))
print(f"Status: {status.values:016b}")
print(f"Bit 0 = {sh.extract_bit(status, 0).values}")
print(f"Bit 2 = {sh.extract_bit(status, 2).values}")
print(f"Bit 3 = {sh.extract_bit(status, 3).values}")
```

```python
status = xr.DataArray(sh.STATUS_TYPE(849))
condition_true = xr.DataArray(True)
condition_false = xr.DataArray(False)
print(f"Status: {status.values:016b}")
print(f"Bit 0 to 1 {sh.set_bit(status, 0, condition_true):016b}")
print(
    f"Bit 0 to 0 {sh.set_bit(status, 0, condition_false, force_zero=True):016b}"
)
print(f"Bit 2 to 1 {sh.set_bit(status, 2, condition_true):016b}")
print(f"Bit 2 to 0 {sh.set_bit(status, 2, condition_false):016b}")
print(f"Bit 0 to 1 {sh.set_bit(status, 0, condition_true):016b}")
print(
    f"Bit 0 to 0 {sh.set_bit(status, 0, condition_false, force_zero=True):016b}"
)
print(f"Bit 2 to 1 {sh.set_bit(status, 2, condition_true):016b}")
print(f"Bit 2 to 0 {sh.set_bit(status, 2, condition_false):016b}")
```

### Extract information from status

```python
status = xr.DataArray(sh.STATUS_TYPE(849))
print(f"Status: {status.values:016b}")
print(
    f"Bit updated value: {sh.extract_bit(status, sh.UPDATED_DATA_BIT_POSITION).values}"
)
print(f"State: {sh.State(sh.extract_state(status).values)}")
print(f"Distance: {sh.extract_distance(status).values}")
```

```python
state = sh.extract_state(status)
xr.testing.assert_equal(
    state, xr.DataArray(sh.State.FORWARD_EXTRAPOLATED.value)
)
```

### Decode status

```python
status = xr.DataArray(sh.STATUS_TYPE(849))
print(f"Status: {status.values:016b}")
updated, state, distance, flags = sh.decode_status(status)
print(
    f"Updated: {updated.values}, State: {sh.State(state.values)}, Distance: {distance.values}, Flags: {flags.values:04b}"
)
```

### Encode status

```python
updated = xr.DataArray(True)
state = xr.DataArray(sh.State.FORWARD_EXTRAPOLATED.value)
distance = xr.DataArray(3)
flags = xr.DataArray(0b0001)
print(
    f"Updated: {updated.values}, State: {sh.State(state.values)}, Distance: {distance.values}, Flags: {flags.values:04b}"
)
encoded = sh.encode_status(updated, state, distance, flags)
print(f"Encoded status: {encoded.values:016b}")
print(f"Encoded status: {encoded.values}")
```
