```python
def bit_array_to_str(bit_array: int, total_bits: int = 16) -> str:
    """
    Convert the bit_array in string format, showing every bit.

    Prameters
    ---------
    bit_array: int
        Integer representing the bit array.
    total_bits: int
        Total number of bits to display (including leading zeros).

    Returns
    -------
    return: str
        String of bit array
    """
    # Format the bit array as a binary string, removing the '0b' prefix
    binary_string = bin(bit_array)[2:]

    # Pad the binary string with leading zeros to match the total_bits
    padded_binary_string = binary_string.zfill(total_bits)

    return str(padded_binary_string)
```
