# Accumulated Precipitations based Model

## General Principle

#### **API**
For an acquisition $t_k:$
$$
\text{API}(t_k) = \lambda \cdot \text{Ratio}(t_k) \cdot 0.2
$$

$$
\text{where} \quad 
\text{Ratio}(t_k) = \frac{\text{ET}(t_k)}{\text{Rad}_{sd}(t_k)}
$$

For a date without acquisition $t_i$:
$$
\text{API}(t_i) = \min\left(k \cdot \text{API}(t_i-1) + P(t_i), \ 100\right)
$$

---
#### **Ratio**
For any date $t$:

$$
\text{Ratio}(t) = \frac{1}{\lambda} \cdot \frac{\text{API}(t)}{0.2}
$$

$$
\text{Ratio}_{smoothed}(t)
= \alpha \cdot \text{Ratio}_{smoothed}(t-1)
+ (1-\alpha) \cdot \text{Ratio}(t),
\quad
\alpha =
\begin{cases}
\alpha_\text{wet} & \text{if } P(t-1) > 0 \\
\alpha_\text{dry} & \text{otherwise}
\end{cases}
$$
---
#### **ET**
For a date without acquisition $t_i:$
$$
\text{ET}(t_i) = \text{Rad}_{sd}(t_i) \cdot \text{Ratio}_{smoothed}(t_i)
$$

For an acquisition $t_k$:
$$
\text{ET}(t_k) = \text{Rad}_{sd}(t_k) \cdot \text{Ratio}(t_k)
$$

**Legend:**

- **$ET$:** daily evapotranspiration
- **$Rad_{sd}$**: daily surface solar radiation  
- **$P$**: daily precipitation  
- **$\lambda = 2,45×10^{-9} J\cdotp kg^{−1}$**: latent heat of vaporization

**Zone-specific parameters:**

- $k$ : scaling parameter
- $\alpha_\text{wet}$: smoothing parameter used when precipitation occurred the previous day
- $\alpha_\text{dry}$: smoothing parameter used when no precipitation occurred the previous day

## Test

We obtained the observed evapotranspiration, daily radiation, and daily explanatory variables (precipitation and soil moisture) using the **prepare_et_single_date**, **prepare_daily_radiation**, and **prepare_daily_explanatory** functions from **ET-DATASET**, respectively. ### Niakhar - Central Senegal

```python
input_dir = os.path.join(
    os.environ.get("SWEAT_TEST_DATA_PATH"), "data_validation", "niakhar"
)
dir_sd, dir_rad, dir_var = find_directories(input_dir)
```

```python
min_date = "2024-01-01"
max_date = "2024-12-31"
x = 345700
y = 1590936
et_sd = et_sd_list(dir_sd, min_date, max_date, x, y)
```

```python
et_ts = et_ts_list_ap(
    dir_rad,
    dir_sd,
    dir_var,
    min_date,
    max_date,
    x,
    y,
    k=0.98,
    freq=3,
    lamb=2.45e9,
    alpha_dry=0.7,
    alpha_wet=0.9,
)
```

```python
plot_et_comparison(et_sd, et_ts, min_date, max_date)
```

```python
print_error_metrics(et_sd, et_ts, 3)
```
